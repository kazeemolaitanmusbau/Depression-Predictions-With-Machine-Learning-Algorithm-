### 1. **Chi-Square Test (For Categorical Features)**
   - **Purpose**: To test the independence between categorical features and the target variable (depression status).
   - **How it Works**: The Chi-Square test checks if the distribution of the target variable is significantly different across the categories of each feature.
   - **Interpretation**: A low p-value (typically < 0.05) suggests that there is a significant association between the feature and the target variable.
 

```python
from scipy.stats import chi2_contingency
categorical_columns = ["country", "Education", "Employed", "Disability", "Mental Health issue", "social media", 
                       "Hour spend on fitness", "wind down", "Mouth dryness","positive feeling"
              ,"breath difficulties", "difficulties in initiate things", "trembling", 
             "worrying situation", "feeling of nothing to look forward to", "down hearted",
             "enthusaistic about anything", "feel meaningless life", "scared"]
for col in categorical_columns:
    contingency_table = pd.crosstab(df[col], df['depression status'])
    chi2, p, dof, expected = chi2_contingency(contingency_table)
    print(f'{col}: p-value = {round(p, 5)}')
```

### 2. **ANOVA (For Numerical Features)**
   - **Purpose**: To compare the means of a numerical feature across the different categories of the target variable (depression status).
   - **How it Works**: ANOVA tests whether the means of the numerical feature are significantly different across the different levels of the categorical target.
   - **Interpretation**: A low p-value indicates that there is a significant difference in the means of the feature across the different depression statuses.

```python
from scipy.stats import f_oneway
numerical_columns = ["Study", "Age"]

for col in numerical_columns:
    groups = [df[col][df['depression status'] == status] for status in df['depression status'].unique()]
    f_stat, p = f_oneway(*groups)
    print(f'{col}: p-value = {p}')
```

The results from the ANOVA test indicate the following:

1. **Study**: p-value = 0.5992  
   - There is **no significant difference** in depression status based on the field of study. The p-value is much higher than 0.05, meaning the study field does not have a noticeable impact on depression status.

2. **Age**: p-value = 0.7389  
   - Age also shows **no significant association** with depression status. The p-value is greater than 0.05, suggesting that depression levels are not significantly different across age groups.

### **Summary**:
Both **study field** and **age** do not show significant differences or associations with depression status based on the ANOVA test results. This implies that neither factor plays a crucial role in determining depression status in this context.
